### Step 6 — ****Run Hypothesis Tests****

🎯 Business Purpose

The goal of this step is to statistically validate whether observed differences and relationships in customer behavior are significant or simply due to random variation.

**Business Question 1:**

Do Male and Female customers spend differently?

Variables
Gender
MonthlySpend

```python
alpha = 0.05
```

```python
# Assumption 1: Normality

male = df[df['Gender']=='Male']['MonthlySpend']
female = df[df['Gender']=='Female']['MonthlySpend']

print(stats.shapiro(male.sample(500)))
print(stats.shapiro(female.sample(500)))
```

```python
# Assumption 2: Equal variance

levene_result = stats.levene(male, female)

print(levene_result)
```

```python
# Running t-test

ttest_result = stats.ttest_ind(
    male,
    female,
    equal_var=False
)

print(ttest_result)
```

```python
# Interpretation

alpha = 0.05

if ttest_result.pvalue < alpha:
    print("Reject H0: There is a significant difference in average MonthlySpend between male and female customers.")
else:
    print("Fail to Reject H0: There is no significant difference in average MonthlySpend between male and female customers.")
```

2. EDUCATION VS MONTHLY SPEND(ONE-WAY ANOVA)

H₀: All means are equal.

H₁: At least one mean differs.

```python
# ---- ASSUMPTION CHECK ---

# Group sizes

print(
    df.groupby('Education')
      ['MonthlySpend']
      .count()
)

print('_____________________________________________________________________')

# Normality

for edu in df['Education'].unique():

    sample = df[
        df['Education']==edu
    ]['MonthlySpend'].dropna()

    print(
        edu,
        stats.shapiro(
            sample.sample(
                min(500,len(sample)) if len(sample) > 0 else 0
            ) if len(sample) > 0 else (np.nan, np.nan) # Handle empty sample after dropna
        )
    )

print('_____________________________________________________________________')

# Homogenity of Variance

groups = [df[df['Education']==edu]['MonthlySpend'].dropna() for edu in df['Education'].unique()]
# Filter out empty groups if any after dropna
groups = [g for g in groups if not g.empty]

# Only run Levene's test if there are at least two groups with data
if len(groups) >= 2:
    print(stats.levene(*groups))
else:
    print("Not enough groups with data to perform Levene's test.")
```

```python
# ANOVA

anova_result = stats.f_oneway(*groups)

print(anova_result)
```

Education levels does not impact average monthly spend.

3 .MARITAL STATUS VS NUMBER OF PETS (CHI-SQUARE TEST)


H₀: Variables are independent.

H₁: Variables are associated.

```python
# Creating contingency table
table = pd.crosstab(df['Married'], df['NumPets'])
print("Contingency Table:")
print(table)

# Chi-Square Test
chi_statistic, p_value, degrees_freedom, expected_frequencies = stats.chi2_contingency(table)

print("\n--- Chi-Square Test Results ---")
print(f"Chi-Square Statistic: {chi_statistic:.2f}")
print(f"P-value: {p_value:.3e}")
print(f"Degrees of Freedom: {degrees_freedom}")

# Interpretation
alpha = 0.05
print("\n--- Interpretation ---")
if p_value < alpha:
    print(f"With p-value ({p_value:.3e}) < alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: Marital status and number of pets are associated.")
else:
    print(f"With p-value ({p_value:.3e}) >= alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: Marital status and number of pets are independent.")
```

Marital status and the number of pets owned are significantly associated.

5.STATE VS MONTHLY SPEND (ONE-WAY ANOVA)



Null Hypothesis (H₀): There is no significant difference in average monthly spending among customers from different states.

Alternative Hypothesis (H₁): At least one state's average monthly spending is significantly different from the others.

```python
# Group counts

print(
    df.groupby('State')
      ['MonthlySpend']
      .count()
)

# Levene test

state_groups = []

for state in df['State'].unique():

    state_groups.append(
        df[df['State']==state]
        ['MonthlySpend']
    )

# ANOVA

state_anova = stats.f_oneway(
    *state_groups
)

print(state_anova)

# Interpretation

alpha = 0.05

if state_anova.pvalue < alpha:
    print("Reject H0: The average monthly spend is significantly different among states.")
else:
    print("Fail to Reject H0: The average monthly spend is not significantly different among states.")
```

### Insight: Monthly Spend by State (One-Way ANOVA)

*   **Test Result:** The One-Way ANOVA yielded a p-value of 0.346, which is greater than the significance level (alpha = 0.05).
*   **Conclusion:** We **fail to reject the null hypothesis**.
